```python
#
import numpy as np

import matplotlib.pyplot as plt

# Read the ECG data from the text file ( this put the data not visualized but in a table format)
ecg_raw = np.loadtxt('data/ECGu.txt')
print(ecg_raw)
# Plot the ECG signal
plt.plot(ecg_raw)
plt.title('ECG') 
plt.xlabel('Time (in ms)')
plt.ylabel('Amplitude (in μV)') 
plt.grid()
plt.show()
```

```python


sampling_frequency=1000 #Frequecy in Hz
T=1/sampling_frequency #Period between 2 samples in s
N=len(ecg_raw)
start=0
stop=(N-1)*T
t= np.linspace(start,stop,N,)

plt.plot(t,ecg_raw)
plt.title('ECG SIGNAL')
plt.xlabel('Time')
plt.ylabel('Amplitude')
plt.grid()
plt.show
```
